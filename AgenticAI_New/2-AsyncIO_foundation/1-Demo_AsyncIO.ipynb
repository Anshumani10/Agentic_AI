Sync vs Async IO

```python
import time
def fetch_data():
   print("fetching data ...")
   time.sleep(5)
   return ("Data received")
def main():
   print(fetch_data())
   print(fetch_data())
main()
```

```python
#Asynchronous way
import asyncio
async def fetch_data():        #Co-routine
   print("fetching data ...")
   await asyncio.sleep(5)
   return ("Data received")
async def main():           #Co-routine
   #print(await fetch_data())
   #print(await fetch_data())
   results = await asyncio.gather(fetch_data(), fetch_data())
   print(results)
await main()
```

```python
import time
def fetch_data_A():
   print("fetching data for A")
   time.sleep(5)
   return ("Data received A")

def fetch_data_B():
   print("fetching data for B")
   time.sleep(5)
   return ("Data received B")

def fetch_data_C():
   print("fetching data for C")
   time.sleep(5)
   return ("Data received C")

def main():
   print(fetch_data_A())
   print(fetch_data_B())
   print(fetch_data_C())
main()
```

```python
# Asynchronous Code MultiFuncation (Non Blocking)
import asyncio
async def fetch_data_A():
   print("fetching data for A ...")
   await asyncio.sleep(5)
   print("fetching data for A!")
   return ("Data received A")

async def fetch_data_B():
   print("fetching data for B ...")
   await asyncio.sleep(5)
   print("fetching data for B!")
   return ("Data received B")

async def fetch_data_C():
   print("fetching data for C ...")
   await asyncio.sleep(5)
   print("fetching data for C!")
   return ("Data received C")

async def main():
   #print(fetch_data_A())
   #print(fetch_data_B())
   #print(fetch_data_C())
   results = await asyncio.gather(fetch_data_A(), fetch_data_B(), fetch_data_C())
   print(results)

await main()
```
