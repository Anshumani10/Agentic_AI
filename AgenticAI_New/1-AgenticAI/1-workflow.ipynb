```python
# Load the API keys into environment variables

from dotenv import load_dotenv

load_dotenv(override = True)
```

```python
#print environment variable
import os
openai_api_key = os.getenv('API_TOKEN')
global deepseek_base_url 
deepseek_base_url = "https://api.deepseek.com"
if openai_api_key:
    print("OpenAI API key exists and begins", openai_api_key[:8])
else:
    print("OpenAI API key is not present, please head to the trouble shooting guide")
    
```

```python
#Start building the workflow and import openai
from openai import OpenAI
from IPython.display import Markdown, display
openai_python_client = OpenAI(api_key=openai_api_key, base_url=deepseek_base_url)
response = openai_python_client.chat.completions.create(
    model="deepseek-v4-pro",
    messages=[
       {"role": "user", "content": "Hello what is 2*2"}
    ],
    stream=False,
)
raw_response = response.choices[0].message.content
display(Markdown(raw_response))
```

```python
#Let's ask question
question = "Please propose a challenging question to assess someone's IQ. Respond only with the question."

messages=[
       {"role": "user", "content": question}
    ]

#Ask the question to deepseek
response = openai_python_client.chat.completions.create(
    model="deepseek-v4-pro",
    messages=messages
)
raw_response_question = response.choices[0].message.content
display(Markdown(raw_response_question))
```

```python
#New question
messages=[
       {"role": "user", "content": raw_response_question}
    ]
```

```python
#Ask it again
response = openai_python_client.chat.completions.create(
    model="deepseek-v4-pro",
    messages=messages
)
raw_response_answer_question = response.choices[0].message.content
display(Markdown(raw_response_answer_question))
```
