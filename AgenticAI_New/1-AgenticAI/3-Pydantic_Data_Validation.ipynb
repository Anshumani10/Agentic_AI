Pydantic data validation examples


```python
from pydantic import BaseModel
#Validate the user data
class User(BaseModel):
    id : int
    name : str

Usr = User(id=1, name="Anshu")
print(Usr)


```

```python
#Use default values
class User(BaseModel):
    id : int
    name : str
    signup_ts : str | None = None
    salary : float | None = None

Usr1 = User(id=2, name = "Peter", salary=40000)
print(Usr1)
```

```python
from dataclasses import dataclass
@dataclass
class User():
    id : int
    name : str
    signup_ts : str | None = None
    salary : float | None = None

Usr1 = User(id=2, name = "Peter", salary=40000)
print(Usr1)
```

Scenario : User Registeration API Input

```python
# You're building a backend service that receives JSON payloads from a front-end form:

{
  "username": "johndoe",
  "password": "secret123",
  "email": "john@example.com",
  "age": "30",
  "join_date": "2025-01-01"
}
```

You need to:

Validate the email format

Ensure age is a positive integer

Ensure password meets minimum length requirements

Convert "30" to integer

Parse "2025-01-01" into a datetime.date

```python
from pydantic import BaseModel, EmailStr, field_validator, ValidationError
from datetime import date

class UserRegisteration(BaseModel):
    username : str
    password : str
    email : EmailStr
    age : int
    join_date : date

@field_validator("age")
@classmethod
def validate_age(cls, value):
    if value <= 0:
        raise ValueError ("Age must be positive")
    return value

@field_validator("password")
@classmethod
def validate_age(cls, value):
    if len(value) < 10:
        raise ValueError ("password must be atleast 10 characters long")
    return value
```

```python
#Simulate incoming json
incoming_data = {
  "username": "johndoe",
  "password": "secret123",
  "email": "john@example.com",
  "age": "30",
  "join_date": "2025-01-01"
}

try :
  user = UserRegisteration(**incoming_data)
  print("JSON OUTPUT :", user.model_dump_json())

except ValidationError as e:
  print(e)
  
```
